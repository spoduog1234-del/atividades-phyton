```python
disciplina=input("Digite o nome da Disciplina: ")
conteudo=input("Digite o conteúdo estudado: ")
atividade=int(input("Digite o número da Atividade: "))
token=disciplina[:3]+conteudo[:2]+str(atividade)+str(atividade**2)+str(atividade//2)+str(atividade+atividade//5)
print("\n---Informações da Atividade---")
print("Disciplina:",disciplina)
print("Conteúdo:",conteudo)
print("Atividade:",atividade)
print("Token:",token)
```

```python
print("English\nPortuguese\nFrench\nRussian\nLatam\nChinese")
language=input("Choose your language: ")
while language!="English" and language!="Portuguese" and language!="French" and language!="Russian" and language!="Latam" and language!="Chinese":
  language=input("Choose your language: ")
if language=="English":
  print("Hello World")
elif language=="Portuguese":
  print("Olá, Mundo")
elif language=="French":
  print("Bonjour le monde")
elif language=="Russian":
  print("Привет, мир")
elif language=="Latam":
  print("Salve, mundo!")
elif language=="Chinese":
  print("你好世界")
else:
  print("Error 404")
```
